```python
import geopandas as gpd
import numpy as np


# 1. Define the path to your .shp file
shapefile_path = "bcalgary_subbasins.shp"

# 2. Read the shapefile
# GeoPandas automatically handles the associated .shx, .dbf, and .prj files!
gdf = gpd.read_file(shapefile_path)


# 2. Initialize the 'is_lakes' column with zeros
gdf['is_lakes'] = 0

# 3. Randomly select 10 unique indices from the DataFrame
# (replace=False ensures we pick 10 distinct rows)
random_indices = gdf.sample(n=10, replace=False, random_state=42).index

# 4. Assign 1 to those sampled indices
gdf.loc[random_indices, 'is_lakes'] = 1

# --- Verify the Results ---
# Check that we have exactly 10 ones and the rest are zeros
print(gdf['is_lakes'].value_counts())

# Take a quick look at a few rows where is_lakes is 1
print(gdf[gdf['is_lakes'] == 1][['COMID', 'is_lakes']].head())

# Save the modified GeoDataFrame to a new shapefile
output_path = "bcalgary_subbasins_lakes.shp"
gdf.to_file(output_path)

print(f"Successfully saved to {output_path}!")
```
